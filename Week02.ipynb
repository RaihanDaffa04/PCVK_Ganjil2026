```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_asli = cv.imread('Foto.jpeg', cv.IMREAD_GRAYSCALE)
tinggi_asli, lebar_asli = img_asli.shape

img_kecil = cv.resize(img_asli, (lebar_asli // 2, tinggi_asli // 2), interpolation=cv.INTER_AREA)

img_rekonstruksi = cv.resize(img_kecil, (lebar_asli, tinggi_asli), interpolation=cv.INTER_CUBIC)


selisih = np.abs(img_asli.astype(np.float32) - img_rekonstruksi.astype(np.float32))
rata_rata_selisih = np.mean(selisih)
piksel_berubah = np.count_nonzero(selisih)
total_piksel = img_asli.size

print(f"Resolusi Asli: {lebar_asli} x {tinggi_asli}")
print(f"Rata-rata selisih intensitas piksel: {rata_rata_selisih:.2f} per piksel (skala 0-255)")
print(f"Jumlah piksel yang nilainya berubah: {piksel_berubah} dari {total_piksel} piksel ({(piksel_berubah/total_piksel)*100:.2f}%)")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_asli, cmap='gray')
plt.title('Citra Asli')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_rekonstruksi, cmap='gray')
plt.title('Citra Hasil Downscale -> Upscale')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def tampilkan(gambar):
    gambar_rgb = cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB)
    plt.imshow(gambar_rgb)
    plt.axis("off")
    plt.show()

def beri_label(path_masukan, teks, path_keluaran):
    img = cv2.imread(path_masukan)
    h, w = img.shape[:2]

    tinggi_bar = h // 12

    kanvas = np.zeros((h + tinggi_bar, w, 3), dtype=np.uint8)
    kanvas[0:h, 0:w] = img
    cv2.rectangle(kanvas, (0, h), (w, h + tinggi_bar), (0, 0, 0), -1)

    skala_font = tinggi_bar / 40
    tebal = tinggi_bar // 25

    x = w // 40
    y = h + tinggi_bar // 2 + tinggi_bar // 4

    cv2.putText(kanvas, teks, (x, y), cv2.FONT_HERSHEY_SIMPLEX,
                skala_font, (255, 255, 255), tebal, cv2.LINE_AA)

    cv2.imwrite(path_keluaran, kanvas)
    tampilkan(kanvas)

beri_label("Foto.jpeg", "Dokumentasi - 2026", "hasil1.jpg")
beri_label("Foto2.jpeg", "Dokumentasi - 2026", "hasil2.jpg")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def tampilkan(gambar):
    gambar_rgb = cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB)
    plt.imshow(gambar_rgb)
    plt.axis("off")
    plt.show()

def buat_thumbnail(path_masukan, nama_toko, path_keluaran):
    img = cv2.imread(path_masukan)
    h, w = img.shape[:2]

    ukuran = max(h, w)
    kanvas = np.full((ukuran, ukuran, 3), 255, dtype=np.uint8)  # kanvas putih

    skala = ukuran / max(h, w)
    baru_w = int(w * skala)
    baru_h = int(h * skala)
    foto_resize = cv2.resize(img, (baru_w, baru_h))

    x = (ukuran - baru_w) // 2
    y = (ukuran - baru_h) // 2
    kanvas[y:y+baru_h, x:x+baru_w] = foto_resize

    skala_font = ukuran / 400
    tebal = ukuran // 200
    cv2.putText(kanvas, nama_toko, (ukuran // 30, ukuran // 15),
                cv2.FONT_HERSHEY_SIMPLEX, skala_font, (0, 0, 0), tebal, cv2.LINE_AA)

    cv2.imwrite(path_keluaran, kanvas)
    tampilkan(kanvas)
    return kanvas

buat_thumbnail("Foto3.jpg", "Toko Sayur", "thumb1.jpg");
buat_thumbnail("Foto4.jpg", "Toko Sayur", "thumb2.jpg");
```
